```python
graph = {
    "Park": {"Primary & Nursery": 1},
    "Primary & Nursery": {"Park":1, "Town Square": 7},
    "Restaurant A": {"Town Square": 3, "Apartment Blocks": 5},
    "Apartment Blocks": {"Town Square": 6, "Hospital": 5, "Restaurant A":5},
    "Town Square": {"Hospital": 5, "Hotel": 6, "Highway": 10, "University":4, "Primary & Nursery":7, "Restaurant A":3, "Apartment Blocks":6},
    "University": {"College": 4, "Student Accoms": 2, "Restaurant B": 3, "Town Square":4},
    "Restaurant B": {"Hospital": 7, "University":3},
    "Highway": {"Hospital": 13},
    "College": {"University":4},
    "Student Accoms": {"University":2},
    "Hotel": {"Town Square":6},
    "Hospital": {"Town Square":5, "Highway":13, "Apartment Blocks":5}
}
# Blocked road: University → Hospital is intentionally NOT included
```

```python
import heapq

def dijkstra(graph, start, goal):
    # Distance table
    distances = {node: float('inf') for node in graph}
    distances[start] = 0

    # Predecessor table
    predecessors = {node: None for node in graph}

    # Priority queue
    pq = [(0, start)]  # (cost, node)

    print(f"\nStarting Dijkstra from {start} to {goal}\n")

    while pq:
        current_cost, current_node = heapq.heappop(pq)

        print(f"Visiting: {current_node} (current cost: {current_cost})")

        # Stop early if goal reached
        if current_node == goal:
            print("\nGoal reached. Stopping search.\n")
            break

        # Explore neighbours
        for neighbour, weight in graph[current_node].items():
            new_cost = current_cost + weight

            print(f"  Checking neighbour: {neighbour} (edge cost: {weight})")

            if new_cost < distances[neighbour]:
                print(f"    Updating: {neighbour} → new cost {new_cost}")
                distances[neighbour] = new_cost
                predecessors[neighbour] = current_node
                heapq.heappush(pq, (new_cost, neighbour))
            else:
                print(f"    No update needed for {neighbour}")

    # Reconstruct path
    path = []
    node = goal
    while node is not None:
        path.append(node)
        node = predecessors[node]
    path.reverse()

    return path, distances[goal]

def run_route_finder():
    print("Available locations:")
    for location in graph.keys():
        print(" -", location)

    start = input("\nEnter your START location: ").strip()
    goal = input("Enter your GOAL location: ").strip()

    # Validate input
    if start not in graph:
        print(f"\nError: '{start}' is not a valid location.")
        return
    if goal not in graph:
        print(f"\nError: '{goal}' is not a valid location.")
        return

    # Run Dijkstra
    path, cost = dijkstra(graph, start, goal)

    print("Final path:", " → ".join(path))
    print("Total cost:", cost)

run_route_finder()
```
